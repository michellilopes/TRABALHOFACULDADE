```python
import pandas as pd
base_census = pd.read_csv("census.csv") #Importando as bases de dados
base_census
```

```python
colunas = base_census.columns[:-1] # selecionando todas as colunas menos a ultima
colunas
```

```python
X_census = base_census.iloc[:,0:14].values

y_census = base_census.iloc[:, 14].values
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder_workclass = LabelEncoder()
label_encoder_education = LabelEncoder()
label_encoder_marital = LabelEncoder()
label_encoder_occupation = LabelEncoder()
label_encoder_relationship = LabelEncoder()
label_encoder_race = LabelEncoder()
label_encoder_sex = LabelEncoder()
label_encoder_country = LabelEncoder()

X_census[:,1] = label_encoder_workclass.fit_transform(X_census[:,1]) # convertendo para numero o valor categorico com label encode
X_census[:,3] = label_encoder_education.fit_transform(X_census[:,3])
X_census[:,5] = label_encoder_marital.fit_transform(X_census[:,5])
X_census[:,6] = label_encoder_occupation.fit_transform(X_census[:,6])
X_census[:,7] = label_encoder_relationship.fit_transform(X_census[:,7])
X_census[:,8] = label_encoder_race.fit_transform(X_census[:,8])
X_census[:,9] = label_encoder_sex.fit_transform(X_census[:,9])
X_census[:,13] = label_encoder_country.fit_transform(X_census[:,13])
X_census
```

```python
from sklearn.preprocessing import MinMaxScaler   #Aplicando normalização dos dados para que fiquem todos na mesma faixa....
scaler = MinMaxScaler()
X_census_scaler = scaler.fit_transform(X_census)
X_census_scaler
```

```python
X_census[:,0].var(), X_census_scaler[:,0].var() #Calculando e exibindo as variancias
```

```python
from sklearn.feature_selection import VarianceThreshold #importando a bibloteca de seleção
```

```python
X_census_scaler.shape
```

```python
selecao = VarianceThreshold(threshold=0.05)                 #definindo meu valor do limiar(corte), valores menor que o limiar, serao removidos
X_census_variancia = selecao.fit_transform(X_census_scaler) #realizando o filtro
X_census_variancia.shape
```

```python
selecao.variances_  # mostrando o calculo das variances do conjunto de dados
```

```python
import numpy as np
indices = np.where(selecao.variances_ >= 0.05 )  #exibindo os indices das colunas selecionadas
indices
```

```python
colunas_selecionadas = colunas[indices]

colunas_selecionadas
```

```python
# Mantendo automaticamente apenas as colunas selecionadas
# + a variável alvo income
bases_census_variancia = base_census[
    list(colunas_selecionadas) + ['income']
]

bases_census_variancia
```

```python
X_census_variancia = bases_census_variancia.iloc[:, 0:5].values  #selecionando os elementos da variaveis X
y_census_variancia = bases_census_variancia.iloc[:, 5].values    #selecionando os elementos da variavel y
```

```python
X_census_variancia[:,0] = label_encoder_education.fit_transform(X_census_variancia[:,0])     #aplicando label encoder na nova base
X_census_variancia[:,1] = label_encoder_marital.fit_transform(X_census_variancia[:,1])
X_census_variancia[:,2] = label_encoder_occupation.fit_transform(X_census_variancia[:,2])
X_census_variancia[:,3] = label_encoder_relationship.fit_transform(X_census_variancia[:,3])
X_census_variancia[:,4] = label_encoder_sex.fit_transform(X_census_variancia[:,4])

X_census_variancia
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

onehotencoder = ColumnTransformer(transformers=[('OneHot', OneHotEncoder(),
                                                 [0,1,2,3,4])], remainder='passthrough')  #converti para combinação de 0s e 1.
X_census_variancia = onehotencoder.fit_transform(X_census_variancia).toarray()   #Aplicando o OneHot
```

```python
scaler = MinMaxScaler()
X_census_variancia = scaler.fit_transform(X_census_variancia)
```

```python
from sklearn.model_selection import train_test_split             #separando os dados de treinamento
x_census_treinamento_var, X_census_teste_var, y_census_treinamento_var, y_census_teste_var = train_test_split(X_census_variancia, y_census_variancia, test_size=0.15, random_state=0)
```

```python
from sklearn.ensemble import RandomForestClassifier
random = RandomForestClassifier(criterion='entropy',                 #Aplicando treinamento com Random Forest
                                min_samples_leaf=1, min_samples_split=5, n_estimators=100)
random.fit(x_census_treinamento_var, y_census_treinamento_var)
```
